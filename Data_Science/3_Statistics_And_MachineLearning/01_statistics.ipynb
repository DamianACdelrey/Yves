#### Practical Statistics

##### Mean and Median

- **Mean**: The arithmetic average, obtained by adding up all the values in a dataset and dividing by the total number of values. The mean is sensitive to extreme values (outliers), which can skew the result.

- **Median**: The middle value in a sorted dataset. If the number of values is odd, it is the central one; if even, it is the average of the two central values. The median is less affected by outliers and provides a better measure of central tendency for skewed distributions.

- Both mean and median are measures of central tendency, but they can give very different insights depending on the distribution of the data.

```python
import statistics as stats

scores = [64, 68, 71, 72, 73, 75, 76,
78, 80, 81, 82, 84, 85, 87, 92]

mean_score = stats.fmean(scores)
meadian_score = stats.median(scores)

print(f"Mean score: {mean_score} \nMedian score: {meadian_score}")
```

#### Spread: Range and Standard Deviation

The spread of a dataset describes how dispersed the values are.

- **Range**: The range is calculated as the maximum value minus the minimum value in a dataset. It gives a rough measure of the overall spread, but it can be heavily influenced by outliers.

- **Standard deviation**: The standard deviation measures the average distance of each data point from the mean of the dataset. A low standard deviation indicates that the data points are clustered close to the mean, while a high standard deviation indicates that the data are spread out over a wider range. Standard deviation is preferred over the range for measuring spread because it takes into account every value in the dataset, not just the extreme values. In normally distributed data, about 68% of the values fall within one standard deviation of the mean.

```python
min_score = min(scores)
max_score = max(scores)

value_range = max_score - min_score

std_dev = stats.pstdev(scores)

print(f"Maximum score: {max_score} \nMinimum score: {min_score}")
print(f"Range: {value_range} \nStandard deviation: {std_dev}")
```

#### Distribution Shape and Outliers

- **Distribution shape**: The shape of a distribution refers to how the values in a dataset are spread out or clustered. This includes aspects such as whether the data are symmetric or skewed, unimodal or multimodal, and the presence of any long tails. Visualizing the distribution, for example with a histogram or boxplot, can show where values cluster, where they become sparse, and whether outliers exist. Outliers are extreme values that differ significantly from most other data points and can have a strong effect on summary statistics such as the mean and standard deviation. Understanding the shape and outliers of a distribution is crucial for choosing appropriate statistical methods and accurately interpreting data.

**Glossary:**
 
- **Distribution shape**: Describes how values in a dataset are arranged or clustered — e.g., symmetric, skewed, unimodal (single peak), or multimodal (multiple peaks).
- **Symmetric distribution**: A distribution where the left and right sides are mirror images; mean and median are generally similar.
- **Skewed distribution**: A distribution where one tail is longer than the other. If the tail is to the right, it’s "right-skewed" (positive); if to the left, "left-skewed" (negative).
- **Unimodal**: A distribution with a single peak (mode).
- **Multimodal**: A distribution with two or more distinct peaks.
- **Long tails**: Regions of the distribution where relatively few data points are found far from the center.
- **Histogram**: A bar graph that shows the frequency of values in intervals (bins) of the data, useful for visualizing distribution shape.
- **Boxplot**: A graphical summary that shows the median, quartiles, and potential outliers of a dataset.
- **Outlier**: An extreme value that differs greatly from the rest of the data. Outliers can strongly affect summary statistics such as the mean and standard deviation.
- **Summary statistics**: Numeric values (like mean, median, range, and standard deviation) that describe and summarize key characteristics of a dataset.

Many analytical questions focus on the relationships between variables rather than just examining single columns. 
**Correlation** is a statistical measure that describes the strength and direction of a linear relationship between two variables. 

- A positive correlation indicates that as one variable increases, the other tends to increase as well. 
- A negative correlation means that as one variable increases, the other tends to decrease.
- When the correlation is close to zero, it suggests that there is no linear relationship between the variables.

It is important to note that correlation does not imply causation; just because two variables are correlated does not mean that changes in one cause changes in the other. Additionally, correlation measures only linear relationships and may not detect more complex or non-linear associations between variables.

#### Understanding Percentiles

A **percentile** is a measure that indicates the value below which a given percentage of observations in a dataset falls. For example, the 25th percentile (also known as the first quartile, or Q1) is the value below which 25% of the data are found, while the 90th percentile is the value below which 90% of the data fall.

Percentiles are useful for understanding the spread and relative standing of values within a dataset. They help answer questions like:
- *How does a particular score compare to the rest of the data?*
- *What value marks the top 10% of achievers?*

**Common percentiles:**
- **25th percentile (Q1):** Separates the lowest 25% from the rest.
- **50th percentile (Median, Q2):** Divides the data into two equal halves.
- **75th percentile (Q3):** Separates the lowest 75% from the highest 25%.

Visualizations such as boxplots use percentiles to summarize and display the distribution of data, highlighting the range, median, quartiles, and potential outliers.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a small synthetic dataset (e.g., test scores between 0 and 100)
np.random.seed(42)  # for reproducibility
scores = np.random.randint(50, 100, size=30)

# Compute summary statistics
mean_score = np.mean(scores)
median_score = np.median(scores)
std_score = np.std(scores)
percentiles = np.percentile(scores, [25, 50, 75, 90])

print("Scores:", scores)
print(f"Mean: {mean_score:.2f}")
print(f"Median: {median_score:.2f}")
print(f"Standard Deviation: {std_score:.2f}")
print(f"25th, 50th, 75th, 90th percentiles: {percentiles}")

# Plot a boxplot to summarize the distribution, highlighting range, median, quartiles, and outliers
plt.figure(figsize=(7, 4))
plt.boxplot(scores, vert=False, patch_artist=True, boxprops=dict(facecolor='skyblue', color='black'),
            medianprops=dict(color='red'),
            whiskerprops=dict(color='black'),
            capprops=dict(color='black'),
            flierprops=dict(marker='o', markerfacecolor='orange', markersize=8, linestyle='none', markeredgecolor='gray'))
plt.title("Boxplot of Synthetic Test Scores")
plt.xlabel("Score")
# Add percentile lines for clarity (25th, 50th, 75th, and 90th)
for p in percentiles:
    plt.axvline(p, linestyle='--', color='gray', alpha=0.7)
plt.show()
```

The previous cell generated a set of synthetic test scores and calculated several descriptive statistics, including the mean, median, standard deviation, and key percentiles (25th, 50th, 75th, and 90th). It also displayed a boxplot that visually summarizes the distribution of the scores. 

From the results, the mean provides the average score, while the median offers the middle value of the dataset, showing how scores are centered. The standard deviation measures the spread, indicating how much the scores vary around the mean. The percentiles help identify the values below which a certain percentage of the data falls: for example, the 25th percentile highlights the lower end of scores, and the 90th percentile shows the value above which only the top 10% of scores lie.

The boxplot visually depicts these statistics, with the box representing the interquartile range (from the 25th to the 75th percentile), a line indicating the median, and potential outliers shown as individual points. This plot makes it easy to see the overall distribution, spot any unusual values, and compare the central tendency with the spread of the data.
