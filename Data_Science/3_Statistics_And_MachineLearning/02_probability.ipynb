#### This notebook covers some basic topics in probability.

Coin flip example

```python
import random

def ten_flips_heads():
    """Return the number of heads in 10 flips of a fair coin."""
    heads = 0
    for i in range(10):
        if random.random() < 0.5:
            heads += 1
    return heads

ten_flips_heads()
```

```python
trials = [ten_flips_heads() for _ in range(1000)]

print(f"min: {min(trials)}, max: {max(trials)}")

print(f"average: {sum(trials) / len(trials)}") # average number of heads per 10 flips
```

- The average number of heads is already close to 5, which is what we expect for a fair coin flipped 10 times.
- The minimum and maximum values reflect the most extreme outcomes observed among all trials.
- With more trials, the average would converge even closer to the expected value of 5.

#### Frequencies and stability

Probabilities can be interpreted as long-run relative frequencies. In code, this means that if we simulate more and more trials, the proportion of a particular outcome should settle near a
stable value. (Law of large numbers)

```python
def estimate_fice_heads(num_trials):
    count_five = 0
    for i in range(num_trials):
        if ten_flips_heads() == 5:
            count_five += 1
    return count_five / num_trials

print(f"estimate_fice_heads(100): {estimate_fice_heads(100)}")
print(f"estimate_fice_heads(1000): {estimate_fice_heads(1000)}")
print(f"estimate_fice_heads(10000): {estimate_fice_heads(10000)}")
print(f"estimate_fice_heads(100000): {estimate_fice_heads(100000)}")
print(f"estimate_fice_heads(1000000): {estimate_fice_heads(1000000)}")
```

This code estimates the probability of getting exactly 5 heads when flipping a fair coin 10 times.

Statistically, this probability is about 25%. In probability theory, it is calculated using the binomial distribution.
That is: "If I flip a coin 10 times, how likely is it to get exactly 5 heads?" The answer is close to 25%.

The function simulates 10 coin flips, many times, and counts how often 5 heads appear. The fraction of such cases approximates the probability.

The output shows that as we increase the number of simulated trials, our estimate for the probability of 5 heads gets closer to the true value (around 25%). 
This demonstrates the law of large numbers: with more experiments, the observed frequency stabilizes near the actual probability.

#### Reproducibility
 
Set a random seed so that the generated random sequence can be reproduced.

This allows your experiments and results to be repeatable: running the notebook multiple times with the same seed will yield the same outcomes.

```python
random.seed(42)

[ten_flips_heads() for _ in range(10)]
```

```python
def simulate_ten_flip_counts(num_trials=10000, seed=42):
    random.seed(seed)
    return [ten_flips_heads() for _ in range(num_trials)]
```

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

def coin_heads_histogram(
    num_trials: int = 10_000,
    seed: int = 42,
    output_name: str = "fig_coin_heads_hist.pdf",
) -> Path:
    counts = simulate_ten_flip_counts(num_trials=num_trials, seed=seed)
    figures_dir = Path("figures")
    figures_dir.mkdir(parents=True, exist_ok=True)
    output_path = figures_dir / output_name

    plt.style.use("seaborn-v0_8")
    fig, ax = plt.subplots(figsize=(6.0, 4.0))
    ax.hist(
        counts,
        bins=range(0, 12),
        align="left",
        edgecolor="black",
        alpha=0.85,
    )
    ax.set_xlabel("Number of heads in 10 flips")
    ax.set_ylabel("Number of experiments")
    ax.set_title("Distribution of Heads Counts in Ten Coin Flips")
    ax.grid(axis="y", alpha=0.35, linestyle="--", linewidth=0.6)
    fig.tight_layout()
    fig.savefig(output_path, format="pdf")
    plt.close(fig)
    return output_path

coin_heads_histogram()
```
