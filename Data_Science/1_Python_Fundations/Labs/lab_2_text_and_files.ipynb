#### Lab 2: Text and Files

- The goal of this notebook is to process a text file by reading, cleaning, summarizing, and saving the summary.

```python
from pathlib import Path

# 'Data' está en una carpeta superior
data_folder = Path("..") / "Data"
messages_file = data_folder / "lab_2.md"

text = messages_file.read_text(encoding="utf-8")
print(text)
```

```python
lines = text.splitlines()
print(f"Number of lines: {len(lines)}")
print(f"First line: {lines[0]}")
```

```python
def count_messages(lines):
    return len(lines)

def count_by_sender(lines):
    counts = {}
    for line in lines:
        sender = line.split(",", 1)[0]
        counts[sender] = counts.get(sender, 0) + 1
    return counts
```

```python
total = count_messages(lines)
by_sender = count_by_sender(lines)

print(f"Total messages: {total}")
print(f"Messages by sender:")
for sender, count in sorted(by_sender.items()):
    print(f"{sender}: {count}")
```

```python
summary_lines = []
summary_lines.append(f"Total messages: {total}")
summary_lines.append("")
summary_lines.append("Messages by sender:")
for sender, count in sorted(by_sender.items()):
    summary_lines.append(f" - {sender}: {count}")
summary_text = "\n".join(summary_lines)

summary_path = data_folder / "lab_2_summary.md"
summary_path.write_text(summary_text, encoding="utf-8")

print(f"Summary saved to {summary_path}")
```
