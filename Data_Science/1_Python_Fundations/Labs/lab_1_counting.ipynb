```python
events = [
"train:delay",
"train:on_time",
"hotel:check_in",
"train:delay",
"snack:coffee",
"train:on_time",
"hotel:check_out",
"snack:tea",
"train:delay",
]
```

Explanation:

- entry.split(':', 1)[0]:
  This code splits the string `entry` into a list using the colon ':' as a separator.
  The `1` means it will only split at the first colon, so you get two parts:
  for example, "train:delay" becomes ["train", "delay"].
  You can then use [0] to get the first part ("train").

- counts.get(category, 0) + 1:
  This code tries to get the value associated with the key `category` in the dictionary `counts`.
  If the key does not exist yet, it returns `0` instead of causing an error.
  This is helpful for counting, since you can always add 1 to the previous value (or 0 if not found).

```python
def count_by_categories(entries):
    counts = {}
    for entry in entries:
        category = entry.split(':', 1)[0]
        counts[category] = counts.get(category, 0) + 1
    return counts

counts = count_by_categories(events)
print(counts)
```

- Comprehensions help you spot patterns quickly.

```python
categories = [entry.split(':', 1)[0] for entry in events]
print(categories)

train_flags = [c == 'train' for c in categories]
print(train_flags)
```

```python
more_events = [
"walk:city_center",
"walk:park",
"photo:landmark",
"photo:cafe",
"train:on_time",
]

print(count_by_categories(more_events))
```
