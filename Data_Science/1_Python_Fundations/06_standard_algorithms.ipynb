#### This notebook introduces some standard algorithms as recurring patterns for solving common problems, such as:

• finding items

##### Linear search: 

- Look at each item in order until the target is found or the collection ends.

```python
tools = ['Python', 'R', 'SQL', 'Git', 'Bash']
target = 'Git'
fount_at = None

for idx, tool in enumerate(tools):
    if tool == target:
        found_at = idx
        break

print(f"Found {target} at index {found_at}.")
```

##### Counting Frequencies:
- How many times an item appears.
- For each word, get its current count (default to 0 if it's not in the dictionary), add 1, and update the count.

```python
words = ['Python', 'R', 'SQL', 'Python', 'Java', 'Python', 'C++', 'R']
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

##### Sorting

```python
scores = [5, 8, 3, 9, 4]
sorted_scores = sorted(scores)

print(sorted_scores)

rows = [('Adam', 29), ('John', 31), ('Anne', 23), ('John', 30)]
sorted_rows = sorted(rows, key = lambda item:item[1], reverse = True)

print(sorted_rows)
```

##### Binary search:
- If the data is sorted, binary search allows you to efficiently find an item by repeatedly dividing the dataset in half and checking which half the item is in.

```python
values = [14, 23, 37, 41, 56, 78, 91, 102, 300, 10, 76, 32, 27, 19, 100]
values.sort()
print(values)

target = 23

left, right = 0, len(values) - 1
mid = (left + right) // 2

while left <= right and values[mid] != target:
    if target < values[mid]:
        right = mid - 1
    else:
        left = mid + 1
    mid = (left + right) // 2

print(f"Found {target} at index {mid}.")
```

##### Lookup tables with dictionaries:
- Instead of scanning repeatedly, the code can often prepare a mapping once and then
perform direct lookup.
- Lists are natural for ordered scans.
- Dictionaries are natural for named lookup.

```python
hours = {'python': 10, 'r': 12, 'sql': 15} # Direct mapping of names to values
hours['sql'] # direct lookup
```

##### Runtime intuition:
- The time taken by a linear scan (such as a for-loop over a list) increases proportionally with the number of items in the dataset—this is called O(n) runtime, and it means that if you double the data, the operation will take roughly twice as long.
- Dictionary (hash table) lookups, on the other hand, provide near-instantaneous access to values once the mapping structure exists. Technically, dictionary lookups are O(1) on average, regardless of the size of the dictionary, making them extremely efficient for repeated queries.
- Sorting a collection, such as with the list sort() method, requires significant initial computation—typically O(n log n) time—but this preprocessing can enable much faster subsequent operations, like binary search or merging, which would be impractical on unsorted data.
- In practice, understanding these trade-offs between upfront costs (like sorting or building a mapping) and query efficiency (like direct lookup or fast search) is crucial to writing performant code, especially when working with large datasets or in scenarios where the same query may be performed multiple times.

```python
# Example_1

exts = ['csv', 'json', 'txt', 'csv', 'ipynb', 'txt', 'csv']
counts = {}

for ext in exts:
    counts[ext] = counts.get(ext, 0) + 1

sorted_exts = sorted(counts.items(), key = lambda item:item[1], reverse = True) # Items() to convert the dictionary to a list of tuples
print(sorted_exts)
```

Example_2: Suppose a folder scan produced file sizes. A reader wants to know whether a specific size exists,
how often sizes repeat, and what the largest values are.

```python
sizes = [120, 85, 120, 300, 85, 410]
print(300 in sizes)

size_counts = {}
for size in sizes:
    size_counts[size] = size_counts.get(size, 0) + 1

print(f'size_counts: {sorted(size_counts.items())}')

print(f'3 largest values: {sorted(sizes, reverse = True)[:3]}') 
```
