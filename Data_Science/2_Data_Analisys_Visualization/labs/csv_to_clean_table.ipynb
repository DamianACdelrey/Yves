### Data Cleaning Steps

These cleaning steps ensure consistent and reliable analysis.  
We use pandas string, datetime, and numeric conversion methods to standardize column formats (e.g., city names, dates, and numeric values), handle missing or invalid data (`errors="coerce"` replaces wrong values with NaN), and remove duplicate rows with `drop_duplicates()`. This improves data quality and prevents errors in downstream analysis.

```python
trips["city"] = trips["city"].str.strip().str.title()	

trips["date"] = pd.to_datetime(
    trips["date"], 
    errors="coerce", 
    dayfirst=False
)

trips["days_spent"] = pd.to_numeric(trips["days_spent"], errors="coerce")
trips["budget_eur"] = pd.to_numeric(trips["budget_eur"], errors="coerce")

trips = trips.drop_duplicates()
```
