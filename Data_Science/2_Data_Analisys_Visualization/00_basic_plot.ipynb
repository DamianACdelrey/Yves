#### Introduction

In this chapter, we will explore how to transform raw Python objects into insightful analytical stories. Along the way, we will introduce essential data analysis tools such as NumPy, pandas, and plotting libraries. You will learn how to use these tools to identify trends, detect outliers, and understand data distributions such as skewness.

The process can be summarized as follows:

Python objects → arrays and tables → charts and summaries → analytical story

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("seaborn-v0_8") # Consistent style

months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]) # x-axis labels
signups = np.array([42, 48, 51, 57, 61, 66, 70, 65, 72, 76, 68, 80]) # y-axis data

fig, ax = plt.subplots(figsize=(6.6, 3.8)) # Create a figure and axes
ax.plot(months, signups, marker="o", linewidth=2.0, color="black") # Line chart
ax.set_title("Monthly Signups")
ax.set_xlabel("Month")
ax.set_ylabel("Number of Signups")
ax.grid(
    alpha=0.45,
    linestyle="--",
    linewidth=0.6,
)
fig.tight_layout() # Reduce label overlap
fig.savefig("Figures/monthly_signups.pdf")
```

The plot makes the overall trend much clearer than the raw numbers.
