#### Arrays
- Arrays require a new way of thinking: instead of processing one element at a time as in a loop, you work with entire columns (vectors) at once.

```python
import numpy as np

signups = np.array([42, 48, 51, 57, 61, 66, 70, 65, 72, 76, 68])

signups
```

First, inspect the structure

```python
print(signups.shape)
print(signups.dtype)
```

Arrays make it possible to apply transformations to the entire vector at once.

```python
print(signups)
print(signups + 5)
print(signups * 2)
```

```python
values = [42, 45, 32, 12] # list
double = []
for value in values:
    double.append(value * 2)
print(f'double list: {double}')

values = np.array([42, 45, 32, 12])
double = values * 2
print(f'double array: {double}')
```

#### Summaries

```python
print(f'sum(): {signups.sum()} sum of all the elements')
print(f'mean(): {signups.mean()} average of all the elements') 
print(f'std(): {signups.std()} standard deviation of all the elements') 
print(f'var(): {signups.var()} variance of all the elements') 
print(f'max(): {signups.max()} maximum value') 
print(f'min(): {signups.min()} minimum value')
```

#### Boolean Mask
This pattern turns a question about each element into a vector True/False

```python
signups >= 55
```

```python
print(signups)
print(signups[signups >= 70])
```

#### Distribution View

```python
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8") # Plot style
scores = np.array([42, 48, 51, 57, 61, 66, 70, 65, 72, 76, 68])

fig, ax = plt.subplots(figsize=(6.6, 3.8)) # Create a figure and an axis.
ax.hist(
    scores,
    bins=[30, 40, 50, 60, 70, 80, 90], #score intervals
    color="skyblue",
    edgecolor="black",
)
ax.set_xlabel("Score")
ax.set_ylabel("Frequency")
ax.set_title("Distribution of Scores")
ax.grid(
    axis="y",
    alpha=0.45,
    linestyle="--",
    linewidth=0.6,
) # easier bar counting
fig.tight_layout() # reduce the space between the title and the plot
plt.show()
```

A typical analytical pattern with arrays:

- Create an array
- Summarize the data (e.g., calculate mean, median, etc.)
- Apply a condition or rule to filter the data
- Inspect or count the subset that meets the condition

```python
# Example

scores = np.array([61, 75, 88, 92, 79, 84, 73, 95])

print(scores.mean())

great_scores = scores[scores >= 80]
print(great_scores)

passed_mask = (scores >= 80)
print(f' passed_mask: {passed_mask}\n total passed: {passed_mask.sum()}')
# True = 1 False = 0
```

```python
# Example 2

random_numbers = np.random.randint(0, 100, size=10)

print(f'random_numbers: {random_numbers}')

print(f'random_numbers.shape: {random_numbers.shape}')

print(f'random_numbers.dtype: {random_numbers.dtype}')

print(f'random_numbers.mean(): {random_numbers.mean()}')

print(f'random_numbers > 50: {random_numbers[random_numbers > 50]}')
```
