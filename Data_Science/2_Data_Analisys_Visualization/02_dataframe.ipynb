#### Dataframes

#### Dataframes
- Are a table-like data structure with rows, columns and labels, commonly used to organize and analyze data in Python.

#### Series
- Is one column of the dataframe with its label

```python
import pandas as pd

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "channel":["web", "web", "partner", "web", "partner", "web"],
    "signups":[42, 48, 51, 57, 61, 66],
    "campaign":["winter", "winter", "referral", "spring", "referral", "spring"]
})

df
```

#### Inspection
The first habit to develop is to inspect the dataframes before starting any work.
- number of rows and columns
- labels
- types of the series

```python
print(df.head())
print(f"\nThe dataframe has {df.shape[0]} rows and {df.shape[1]} columns")
print(f"\nThe columns are: {df.columns}") # column labels
print(f"\nThe data types are: {df.dtypes}") # columns data types
```

#### Selecting of columns

```python
signups = df["signups"] # one column -> single brackets
print(signups)

monthly_signups = df[["month", "signups"]] # list of columns -> smaller table
print(monthly_signups)
```

#### Row filtering

```python
# Boolean mask: True for rows where signups >= 55
df["signups"] >= 55
```

```python
# keep only rows where signups >= 55
df[df["signups"] >= 55]
```

#### Summaries

- `.describe()` give a compact first pass over the data

```python
print(f"The mean of signups is {df['signups'].mean()}")
print(f'\ncount of channels: {df["channel"].value_counts()}')
print(f'\nGeneral description of the dataframe: \n{df.describe(include="all")}')
```

#### Sorting

```python
df.sort_values(by="signups", ascending=False)
```

#### Example:

- inspect the table,
- choose a column of interest,
- apply a rule,
- inspect the subset,
- summarize what seems worth pursuing next.

```python
subset = df[df["signups"] >= 50]

print(subset[["month", "campaign", "signups"]])
print(f'--' * 30)
print(subset["campaign"].value_counts())
```
