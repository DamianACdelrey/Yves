**Python fundamentals — types and collections**

*Section 1 — Data types*

This notebook introduces core types (strings, numbers, booleans) and why checking `type()` helps avoid subtle bugs.

**Variables.** Choose clear names so you can see what each value represents without extra context.

```python
name = "Ada" #String
hours = 8 #Integer
is_ready = True #Boolean

print(f"Hello, {name}! You have been coding for {hours} hours. Are you ready? {is_ready}")
```

**Expressions.** They combine values and operators to produce new values (for example, `2 + 3` or `str(x)`).

```python
hours_monday = 2
hours_tuesday = 3
total_hours = hours_monday + hours_tuesday

message = "Total hours: " + str(total_hours)

print(message)
```

**Strings.** Numbers often represent amounts; strings are text *labels*. In practice you need both: numeric values to compute with, and text to say what those values mean.

```python
title = "Python fundamentals"
platform = "Cursor"

title + " on " + platform
```

**Numbers.** Integers (`int`) and floating-point values (`float`) cover most day-to-day work; more detail (precision, rounding) comes later.

```python
hours = 3 # int
total_cost = 25.5 # float

average_cost = total_cost / hours

print(f"The average cost per hour is {average_cost}")
```

**Booleans (`True` / `False`).** These values underpin conditional logic—the same ideas as basic logic gates (AND, OR, NOT, and common compositions).

```python
is_complete = False
is_beginner = True

print(f"AND gate -> {is_complete} and {is_beginner} -> {is_complete and is_beginner}")
print(f"OR gate -> {is_complete} or {is_beginner} -> {is_complete or is_beginner}")
print(f"NOT gate -> {is_complete} -> {not is_complete}")
print(f"XOR gate -> {is_complete} != {is_beginner} -> {is_complete != is_beginner}")
print(f"XNOR gate -> {is_complete} == {is_beginner} -> {is_complete == is_beginner}")
print(f"NAND gate -> not ({is_complete} and {is_beginner}) -> {not (is_complete and is_beginner)}")
print(f"NOR gate -> not ({is_complete} or {is_beginner}) -> {not (is_complete or is_beginner)}")
```

**Checking types.** Much confusion in data work comes from assuming a value has one kind of shape when it is actually another. Beginners should use `type()` (and related checks) early and often.

```python
name = "Maria"
type(name)
```

```python
scores = [8, 9, 6]
type(scores)
```

**Section 2 — Collections**

Collections group multiple items in one object (items may have different types).

- **List** — Ordered and mutable. Use when order matters, you need to add or remove items, or you will iterate often.
- **Tuple** — Ordered and immutable. Use when the sequence should not change after creation.
- **Dictionary** — Maps keys to values. Fast lookup by a unique key; order is secondary (though modern Python preserves insertion order).
- **Set** — Unordered collection of unique items. Good for membership tests and deduplication.

```python
# List
scores = [8, 9, 5, 6]
scores.append(10)

print(f"Scores after .append(10): {scores}")

scores.remove(5)

print(f"Scores after .remove(5): {scores}")

scores.sort()

print(f"Sorted scores: {scores}")
```

```python
# Tuples
point = (10, 20)

type(point)
```

```python
point = (10, 20)
rgb = (255, 255, 125)

print(f"First element of point: {point[0]} \nLast element of rgb: {rgb[-1]}")
print(f"Index of first 255 in rgb: {rgb.index(255)}")
print(f"Count of 255 in rgb: {rgb.count(255)}")
```

```python
# Dictionaries
student = {'name': 'Ada', 'track': 'Data science', 'hours_this_week': 5}

print(f"Student name: {student['name']}")
print(f"Student track: {student['track']}")
print(f"Student hours this week: {student['hours_this_week']}")

student.keys()
```

```python
# Sets

tools = {'python', 'git', 'python', 'colab'}

print(tools)
print('python' in tools)
print('sql' in tools)
```

**Indexing and slicing**

*Indexing* picks one position (e.g. `word[0]`). *Slicing* picks a range with `[start:end]` (the end index is exclusive).

```python
word = 'python'

print(word[0])
print(word[1:3])
```

**Worked example.** Combining a dict, a list, and a boolean check.

```python
student = {'name': 'Ada', 'track': 'Data science'}
sessions = [45, 30, 60]

total_minutes = sum(sessions)
is_consistent = total_minutes >= 120

print(f'{student["name"]} has studied {total_minutes} minutes. Consistent? -> {is_consistent}')
```

**Second example.** Dictionary keys, an f-string with mixed quotes, and mutating the dict with `pop`.

```python
book = {'title': 'Advances in Financial Machine Learning',
    'difficulty': 'high',
    'author': 'Marcos Lopez',
    'editor': 'Wiley'}

print(f"The book title is: {book['title']}")

print(book.keys())
book.pop('editor')
book.keys()
```
