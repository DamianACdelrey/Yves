**Control flow and functions**

You will use branches (`if` / `else`), loops (`for` / `while`), and small reusable functions in almost every script.

**Conditionals (`if` / `else`).** Run different code depending on whether a condition is true.

```python
import random
random_number = random.randint(1, 100)

if random_number % 2 == 0:
    label = "even"
else:
    label = "odd"

print(f"The random number {random_number} is {label}")
```

**Loops (`for` / `while`).** `for` walks over a sequence (or iterator); `while` repeats until a condition becomes false.

```python
scores = [7, 8, 12, 4, 9, 10]

# Simple accumulation with for
total = 0
for score in scores:
    total += score
    print(f"The current total is {total}, {score} was added")
print(f"The total score is {total}")

print("-" * 40)

# range(): loop by index
for i in range(len(scores)):
    print(f"Score at position {i} is {scores[i]}")

print("-" * 40)

# enumerate(): index and value together
for idx, score in enumerate(scores):
    print(f"Score #{idx} is {score}")

print("-" * 40)

# zip(): pairs elements until the shorter iterable ends
students = ["Ana", "Luis", "Marta", "Juan", "Sara", "Pedro"]
for name, score in zip(students, scores):
    print(f"{name} scored {score}")

print("-" * 40)

# List comprehension — build a new list in one expression
squared_scores = [score**2 for score in scores]
print(f"Squared scores: {squared_scores}")
```

```python
# while: repetition depends on a condition. If the condition never becomes false, the loop runs forever.
n = 3
steps = []

while n > 0:
    steps.append(n)
    n -= 1
else:
    # else runs when the loop ends without break
    print("Loop finished normally (no break)")

print(steps)
```

**Functions.** Name a piece of logic, pass inputs, return results—reuse the same steps without duplicating code.

**Input** ⟶ **transform** ⟶ **decide** ⟶ **output**

```python
def average_scores(values):
    total = sum(values)
    return total / len(values)
```

```python
scores = [7, 8, 12, 4, 9, 10]
print(average_scores(scores))
```

```python
def is_long_word(word, threshold):
    return len(word) >= threshold
```

```python
word = "supercalifragilistic"
print(is_long_word(word, 10))
```

```python
def weekly_status(session, target = 120):
    total = sum(session)
    status = 'Consistent' if total >= target else 'Inconsistent'
    return total, status

total, status = weekly_status([30, 40, 60])
print(f"Total minutes this week: {total}, Status: {status}")
```

**Exercise.** Write a function that returns both the maximum of a list and how many values it contains.

```python
def largest_and_count(numbers):
    largest = max(numbers)
    count = len(numbers)
    return largest, count

numbers = [1, 300, 5, 7, 9, 11, 13, 15, 17, 19]
largest, count = largest_and_count(numbers)

print(f"The largest number is {largest} and the list has {count} values")
```
