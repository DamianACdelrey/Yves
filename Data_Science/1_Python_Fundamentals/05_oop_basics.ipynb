#### This notebook contains the basics of object-oriented programming.

A class is a blueprint; an object is a concrete instance created from that blueprint.

• The class defines what kind of thing this is.

• The object is one actual thing of that kind.

• Attributes store state.

• Methods define behavior.

```python
class StudySession:
    def __init__(self, topic, minutes): # Constructor
        self.topic = topic # Attribute
        self.minutes = minutes # Attribute

    def is_long(self): # Method/Behavior
        return self.minutes > 30
    
    def summary(self):
        return f"Learning {self.topic} for {self.minutes} minutes"

session = StudySession("Python", 45)
print(session.topic, session.minutes)

session.is_long()
```

The advantage of using a class instead of, for example, a dictionary is that a class provides a clearer and more organized structure, especially when you need to apply the same behaviors or operations repeatedly. With a class, methods are bundled together with the data, making it easier to manage state and functionality in one place.

In a class, the special method `__init__` is called the constructor; it initializes each new object with specific values. The parameter `self` refers to the current object instance, allowing you to set or access object attributes and call methods within the class.

Methods are not floating separately. They are attached to the class because those behaviors belong naturally to the object.

```python
StudySession('git', 30).summary()
```

One class can generate many objects that share structure but carry different state

```python
Ada = StudySession('git', 30)
Lyn = StudySession('python', 10)

print(Ada.summary())
print(Lyn.summary())
```

While classes are powerful, you do not always need them. For simple code that just stores related data temporarily, such as a small script or a calculation, using basic data structures like dictionaries, lists, or tuples might be sufficient.

If your task does not involve repeated behaviors, state management, or code that must scale to many similar objects, then functions and basic containers are often simpler and clearer than creating a class.

```python
class Learner:
    def __init__(self, name, minutes):
        self.name = name
        self.minutes = minutes

    def status(self, target = 120):
        return 'on track' if self.minutes >= target else 'building'
    
    def add_minutes(self, extra):
        self.minutes += extra

learner = Learner('Ada', 100)
print(learner.name, learner.status())

learner.add_minutes(20)
print(learner.name, learner.status())
```

```python
# Exercise

class Transaction:
    def __init__(self, amount, date, description):
        self.amount = amount
        self.date = date
        self.description = description
    
    def summary(self):
        return f"{self.date}: {self.description} - {self.amount}"

transaction = Transaction(100, '2021-01-01', 'Amazon')
print(transaction.summary())
```
