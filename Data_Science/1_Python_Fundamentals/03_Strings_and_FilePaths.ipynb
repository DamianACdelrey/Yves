4. Writing and Reading:

The keyword 'with' is used in Python to simplify working with files. It automatically manages opening and closing the file, even if errors occur during processing.

This makes the file workflow safer and cleaner, as you don’t have to remember to close the file yourself.

The standard pattern is: open the file using 'with', do your work, and Python will handle closing it for you when you're done.

```python
with open("data/output.txt", "w", encoding="utf-8") as writer:
    _ = writer.write("Python\nGit\nColab\nPython")

with open("data/output.txt", "r", encoding="utf-8") as reader:
    text = reader.read()

text
```

```python
lines = text.strip().split("\n") # one line per element
print(lines)

len(lines), len(set(lines)) # total vs unique

# read, inspect, split, summarize
```

```python
# save the results

summary = "Unique tools: " + ", ".join(sorted(set(lines)))

with open("data/summary.txt", "w", encoding="utf-8") as writer:
    _ = writer.write(summary) # save the summary

with open("data/summary.txt", "r", encoding="utf-8") as reader:
    saved = reader.read()

print(saved)
```

5. Python’s built-in pathlib module gives a calmer way to handle paths across operating
systems. The important idea is not the library itself. The important idea is that paths can be built
from meaningful pieces instead of typed out as fragile long strings

```python
from pathlib import Path

data_dir = Path("data") # relative folder
summary_path = data_dir / "summary.txt"

summary_path.exists() # check if the file exists
print(summary_path)
```

```python
# Example

def clean_topics(text):
    parts = text.split("\n") # split into lines
    cleaned = []
    for part in parts:
        item = part.strip().lower() # normalize each line
        if item:
            cleaned.append(item) # keep non-empty items
    return cleaned

clean_topics(" Python \nGit\n\nColab ")
```

```python
# Example 2

def create_and_save_file(path, text):
    with open(path, "w", encoding="utf-8") as writer:
        _ = writer.write(text)

create_and_save_file("data/example_2.txt", "  Python\n  Git\nColab\n Sql ")

with open("data/example_2.txt", "r", encoding="utf-8") as reader:
    text = reader.read()

print(text)

cleaned = clean_topics(text)
print(cleaned)
```
