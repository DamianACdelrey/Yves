5. Booleans. This type is the foundation of computational logic (logic gates).

```python
is_complete = False
is_beginer = True

print(f"AND gate -> {is_complete} and {is_beginer} -> {is_complete and is_beginer}")
print(f"OR gate -> {is_complete} or {is_beginer} -> {is_complete or is_beginer}")
print(f"NOT gate -> {is_complete} -> {not is_complete}")
print(f"XOR gate -> {is_complete} != {is_beginer} -> {is_complete != is_beginer}")
print(f"XNOR gate -> {is_complete} == {is_beginer} -> {is_complete == is_beginer}")
print(f"NAND gate -> not ({is_complete} and {is_beginer}) -> {not (is_complete and is_beginer)}")
print(f"NOR gate -> not ({is_complete} or {is_beginer}) -> {not (is_complete or is_beginer)}")
```

##### 2. Collections — containers for holding multiple items, possibly of different data types.
<span style="color:green"><b>1. List</b></span> — Ordered and mutable. Best when you need to maintain order, allow changes (add/remove items), or expect to iterate repeatedly.

<span style="color:green"><b>2. Tuple</b></span> — Ordered and immutable. Use when the sequence of items should not change after creation.

<span style="color:green"><b>3. Dictionary</b></span> — Key/value mapping. Useful for fast lookup via a unique key; item order is less important.

<span style="color:green"><b>4. Set</b></span> — Unordered collection of unique items. Ideal for membership checks and removing duplicates.

```python
# List
scores = [8, 9, 5, 6]
scores.append(10)

print(f"Scores after .append(10): {scores}")

scores.remove(5)

print(f"Scores after .remove(5): {scores}")

scores.sort()

print(f"Sorted scores: {scores}")
```
