#### What Does It Mean to Think Pythonically?

Writing Pythonic code means creating solutions that are readable, straightforward, expressive without being cryptic, and consistent with established Python practices. Pythonic code is not about showing off or using fancy tricks—it's about clarity and simplicity. The main goal is for your code to communicate its purpose as transparently and efficiently as possible, reducing misunderstandings and maintenance effort.

A practical way to gauge whether your code is Pythonic is to consider how easily another Python programmer can understand it. If your code’s structure and intent are immediately clear to someone reading it for the first time, you are likely on the right path.

Keep in mind: Python values community conventions and clear intent over unnecessary complexity. Writing with this mindset helps you create code that is easy to read, maintain, and collaborate on.

• good names

• small functions

• short comments where they help

• and examples that show both code and result

`Truthiness`:  
A core Pythonic habit is to leverage the "truthiness" of objects when making decisions in code. Instead of using methods like `is_empty()` or explicit length checks, you can let objects naturally determine their truth value in conditions.  

This approach is useful because it makes code shorter, clearer, and more idiomatic. For example, in Python, empty collections (like lists, dictionaries, sets, or strings) automatically evaluate to `False`, while non-empty ones evaluate to `True`. This means you can write simple, readable code without unnecessary boilerplate, making your intentions obvious to other Python programmers. By trusting Python's built-in behavior, you reduce redundancy and improve code maintainability.

```python
items = []

if items:
    status = 'has data'
else:
    status = 'empty'

status
```

`Enumerate:` Common pattern when position and value are needed

```python
tools = ['Python', 'colab', 'git']
pairs=[]

for idx, tool in enumerate(tools, start=1):
    pairs.append((idx, tool))

pairs
```

`FOR loops:` A beginer aproach is to write the plain loop first and then evaluate if an idiom make clearer the result

```python
prices = [12.5, 9, 10, 22.5]
doubled_prices = []

for price in prices:
    doubled_prices.append(price * 2)
print(f"Doubled prices: {doubled_prices}")

doubled_prices = [price * 2 for price in prices] # same transformation more compact
print(f"Doubled prices: {doubled_prices}")
```

`ZIP:`

The `zip` function in Python is used to combine two or more iterables (such as lists or tuples) into a single iterator of tuples. Each tuple contains one element from each iterable, paired together by their position in the original iterables. This is especially useful when you need to process items in parallel or create pairs from separate sequences.

```python
names = ['John', 'Jane', 'Jim']
scores = [95, 87, 88]

names_scores = list(zip(names, scores))
print(f"list of tuples: {names_scores}")

rows=[]

for name, score in zip(names, scores):
    rows.append(f"{name}: {score}") # combine label and value

print(f"list of strings: {rows}")
```

`Comprehensions:` express a common transformation in one place. If a comprehension needs multiple branches, deep nesting, or heavy side effects, it is usually a sign that the plain loop version would be clearer.

```python
nums = [1, 2, 3, 4, 5]

squares = [n * n for n in nums]

print(squares)
```

```python
# bad VS good
raw_names = ['  John', '  Jane', 'Jim  ', '']
labels = [name.strip().title() if name.strip() else 'Unknown' for name in raw_names]

# Easier to read:
labels = []
for name in raw_names:
    cleaned = name.strip()
    if cleaned:
        labels.append(cleaned.title())
    else:
        labels.append('Unknown')

print(labels)
```

`Filtering:`Many data tasks involve keeping some values and discarding others. A comprehension can
express that clearly when the condition is simple.

```python
scores = [80, 50, 60, 70, 85, 88, 55, 40]

passed = [score for score in scores if score >= 60]

print(passed)
```

`Sorting`

```python
names = ['Li', 'Ada', 'Lucas', 'Margot', 'Kelly']
sorted_names = sorted(names)
print(sorted_names)

sorted_by_length = sorted(names, key=len)
print(sorted_by_length)
```

```python
# key based sorting
records = [('Ada', 81), ('John', 70), ('Jane', 91)]
sorted_records = sorted(records, key=lambda item:item[1], reverse=True) # Sort by score
print(sorted_records)
```

```python
# Example
topics = ['pyThon  ', 'Colab ', ' Pandas', ' data', 'pythonic ', 'pYthon']
cleaned = [topic.strip().lower() for topic in topics] # Normalize each item

print(cleaned)

result = sorted(set(cleaned))
print(result)
```

```python
# Example 2
names = [' Ada', 'lin ', 'MAX   ']
sessions = [ 6, 4, 2]

clean_names = [name.strip().title() for name in names]
print(clean_names)

rows = list(zip(clean_names, sessions))
sorted_rows = sorted(rows, key=lambda item:item[1], reverse=True)
print(sorted_rows)
```

`Lambda` is a way to create small, anonymous functions in Python using the `lambda` keyword. Lambdas are useful when you need a simple function for a short period, especially for operations like sorting or filtering data where a full `def` function would be unnecessarily verbose.

In the line:

`sorted_rows = sorted(rows, key=lambda item: item[1], reverse=True)`

- `lambda item: item[1]` creates a function that takes an input (`item`) and returns its second element (`item[1]`).
- `key=lambda item: item[1]` tells `sorted` to use that returned value as the sorting key.
- `reverse=True` sorts from highest to lowest instead of lowest to highest.

So, `sorted` uses the value from the lambda function to determine the order of the list.

```python
# Example 3
tools = ['Python', 'colab', 'git']
pairs=[3, 1, 2]

paired_tools = list(zip(tools, pairs))
sorted_paired_tools = sorted(paired_tools, key=lambda item: item[1])

sorted_paired_tools
```
